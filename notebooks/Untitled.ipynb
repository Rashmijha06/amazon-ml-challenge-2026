```python
import os
import shutil

# 1. Define required directory structure
directories = [
    "dataset/train",
    "dataset/test",
    "output",
    "code/business_entity_resolution/src",
    "utils",
]

for directory in directories:
    os.makedirs(directory, exist_ok=True)
    print(f"Directory ready: {directory}")

# 2. Identify and move uploaded train files into dataset/train/
train_files = [
    "train_source1.tsv",
    "train_source2.tsv",
    "train_source3.tsv",
    "train_ground_truth.tsv",
]

for filename in train_files:
    if os.path.exists(filename):
        target_path = os.path.join("dataset/train", filename)
        shutil.move(filename, target_path)
        print(f"Moved {filename} -> dataset/train/{filename}")
    else:
        print(f"File {filename} not found in root (already moved or missing).")
```

```python
import pandas as pd

# Load train datasets
train_s1 = pd.read_csv("dataset/train/train_source1.tsv", sep="\t")
train_gt = pd.read_csv("dataset/train/train_ground_truth.tsv", sep="\t")

print("Train S1 shape:", train_s1.shape)
print("Ground truth shape:", train_gt.shape)
display(train_s1.head(3))
```

```python
import os

# Files to remove from the root folder since they are already safely in dataset/train/
files_to_remove = [
    "train_source1.tsv",
    "train_source2.tsv",
    "train_source3.tsv",
    "train_ground_truth.tsv",
]

for filename in files_to_remove:
    if os.path.exists(filename):
        os.remove(filename)
        print(f"Removed redundant root file: {filename}")
    else:
        print(f"File not found in root (already removed): {filename}")
```
